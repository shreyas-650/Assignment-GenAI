## Task 2: Hugging Face Emedding Model

```python
from langchain_huggingface import HuggingFaceEmbeddings
```

```python
embedding = HuggingFaceEmbeddings(model='sentence-transformers/all-MiniLM-L6-v2')
```

```python
documents = [
    'My name is Shreyas',
    'This is the example document',
    'Langchain is very popular now a days',
    'Dog,cat,lions are all animals'
]
```

```python
res = embedding.embed_documents(documents)
```
