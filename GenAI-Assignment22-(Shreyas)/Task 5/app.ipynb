## Task 5: Similarity Search with Langchain Abstraction

```python
from langchain_huggingface import HuggingFaceEmbeddings # OpenAI Credits are not there hence using HuggingFaceEmbedding
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
embedding = HuggingFaceEmbeddings(model='sentence-transformers/all-MiniLM-L6-v2')
```

```python
docs = TextLoader('../data/obama.md',encoding='utf-8').load()
splitter = RecursiveCharacterTextSplitter(chunk_size=400,chunk_overlap=0)

chunks = splitter.split_documents(docs)
texts = [text.page_content for text in chunks]
```

```python
len(texts)
```

```python
res = embedding.embed_documents(texts)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
query = embedding.embed_query('Who is the president of america')
scores = cosine_similarity([query],res)
```

```python
scores
```

```python
index,value=sorted(list(enumerate(scores)),key= lambda x:x[1])[-1]
```

```python
print(texts[index])
print('\nThe similarity score is\n',value)
```

```python
from langchain_chroma import Chroma
vector_store = Chroma(embedding_function=embedding,collection_name='chroma_db')
```

```python
vector_store.add_documents(chunks)
```

```python
result = vector_store.similarity_search(query='China',k=5)
```

```python
for docs in result:
    print('\nPage Content\n',docs.page_content)
    print('\nMeta Data\n',docs.metadata)
    print('\nID\n',docs.id)
```
