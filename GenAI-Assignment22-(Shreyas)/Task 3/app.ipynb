## Task 3: Comparison: OpenAI vs Hugging Face Embeddings

1. When to prefer OpenAl embeddings? 
When we have to get more detailed and accurate embedding
2. When to prefer Hugging Face embeddings? 
When we are working in small project or prototype where semantic score is not that much important
3. Cost vs performance trade-offs.
Hugging Face cost low as it runs locally whereas Open AI runs on credits usage (Pay as you used model)
Performance of OpenAI is more accurate as compared to Hugging Face Model.
