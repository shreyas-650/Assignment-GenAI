## Task 4: Similarity Search App(Core Logic)

```python
from langchain_huggingface import HuggingFaceEmbeddings # OpenAI Credits are not there hence using HuggingFaceEmbedding
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_chroma import Chroma
```

```python
embedding = HuggingFaceEmbeddings(model='sentence-transformers/all-MiniLM-L6-v2')
```

```python
docs = TextLoader('../data/obama.md',encoding='utf-8').load()
splitter = RecursiveCharacterTextSplitter(chunk_size=400,chunk_overlap=0)

chunks = splitter.split_documents(docs)
```

```python
vector_store = Chroma(embedding_function=embedding,persist_directory='chroma_db',collection_name='sample_data')
```

```python
vector_store.add_documents(chunks)
```

```python
vector_store.get(include=['documents','embeddings','metadatas'])
```

```python
vector_store.similarity_search(query='Who is president',k=1)
```

```python
def search(queryy,top_k=3):
    res = vector_store.similarity_search(query=queryy,k=top_k)
    return res
```

```python
search('who is president')
```

```python
search('Donal Trump',1)
```

```python
search('India')
```
