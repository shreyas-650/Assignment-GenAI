## Task 7: FAISS Vector Store

```python
from langchain_community.embeddings import OllamaEmbeddings
from langchain_community.vectorstores import FAISS
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
embedding = OllamaEmbeddings(model='nomic-embed-text')
```

```python
docs = TextLoader('../data/obama.md').load()
spliter = RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=0)
chunks = spliter.split_documents(docs)
```

```python
db = FAISS.from_documents(chunks,embedding)
```

```python
result = db.similarity_search(query='Powerful Man',k=5)
```

```python
db.save_local('faiss_index')
```

```python
new_db = db.load_local('faiss_index',embedding,allow_dangerous_deserialization=True)
```

```python
new_db.similarity_search(query='Trump President',k=3)
```
