```python
# Using Groq with Open AI Model since OpenAI API is chargeable.
from dotenv import load_dotenv
from langchain_core.prompts import ChatPromptTemplate,MessagesPlaceholder
from langchain_core.runnables import RunnableWithMessageHistory
from langchain_groq import ChatGroq
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.output_parsers import StrOutputParser
load_dotenv()
store = {}

def get_history(session_id : str):
    if session_id not in store:
        store[session_id]=ChatMessageHistory()
    return store[session_id]

prompt = ChatPromptTemplate([
    ('system','You are a rude assistant which talk in very rude way'),
    MessagesPlaceholder(variable_name='history'),
    ('human','{query}')
])
LLM = ChatGroq(model='openai/gpt-oss-20b')
chain = prompt | LLM | StrOutputParser()

chatbot = RunnableWithMessageHistory(
    chain,
    get_history,
    input_messages_key='query',
    history_messages_key='history'

)
```

```python
chatbot.invoke({'query':'My name is Shreyas and I like to dance in public places, along with that I like to do coding'},
               config={'configurable':{'session_id':'shrey123'}})
chatbot.invoke({'query':'Tell me how to write the basic hello world code in python'},
               config={'configurable':{'session_id':'shrey123'}})
chatbot.invoke({'query':'Do you like what i like?'},
               config={'configurable':{'session_id':'shrey123'}})
chatbot.invoke({'query':'Buy why you like?'},
               config={'configurable':{'session_id':'shrey123'}})
chatbot.invoke({'query':'Tell me what I like to do'},
               config={'configurable':{'session_id':'shrey123'}})
```
