## Task 12: Observations & Insights

### 1. Which loader is used for which data type? 
### For Text - TextLoader
### Text PDF - PyPDFLoader
### Image Text PDF - PyMuPDFLoader
### Scanned Img - UnstructuredPDFLoader
### CSV - CSVLoader
### Web - WebBasedLoader

### 2. Best splitter for: 
### o Small text 
### Character Text Splitter (length based)
### o Large PDFs 
### Text Structure Based (Recursive Character Text Splitter)
### o Web data 
### Document Structure Based (RCTS.from_language)

### 3. Why chunk overlap is important?
### Chunk overlap helps to preserve the text in the length based text splitter as if the text ### got splitted the text will be useless hence overlap is used
