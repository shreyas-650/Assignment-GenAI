## Task 10: Semantic Meaning Based Splitting

1. Explain what semantic chunking means. 
Semantic Chunking is the text splitting technique in which if sentance 1 and sentance 2 have good semantic value there will be in the same chunk and if not seperate chunk will be made.
2. Briefly describe how embeddings help semantic splitting.
Split by creating embedding of a sentence and then those of similar embedding are then grouped together
