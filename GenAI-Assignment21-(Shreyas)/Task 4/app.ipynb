## Task 4: Directory Loader

```python
from langchain_community.document_loaders import DirectoryLoader,PyPDFLoader,TextLoader,CSVLoader
```

```python
pdf_docs = DirectoryLoader('../data',glob='**/*.pdf',loader_cls=PyPDFLoader).load()
csv_docs = DirectoryLoader('../data',glob='**/*.csv',loader_cls=CSVLoader).load()
txt_docs = DirectoryLoader('../data',glob='**/*.txt',loader_cls=TextLoader).load()
docs = pdf_docs + csv_docs + txt_docs
```
