## Task 5: Multi-Query Retriever

```python
from langchain_classic.retrievers import MultiQueryRetriever
from langchain_community.vectorstores import FAISS
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_ollama import ChatOllama
```

```python
embedding = HuggingFaceEmbeddings(model='sentence-transformers/all-MiniLM-L6-v2')
```

```python
LLM = ChatOllama(model='llama3.2',temperature=0.5)
```

```python
docs = [ "Deep Learning uses neural networks with many layers to learn complex patterns.",
        "Generative AI uses Deep Learning models like GANs and Transformers to create new content.", 
        "The breakthrough in Generative AI came from large-scale training of deep neural networks.", 
        "Transformers, a Deep Learning architecture, enabled large language models like GPT.", 
        "The Eiffel Tower is located in Paris, France.", 
        "Photosynthesis is the process by which plants make food using sunlight.", 
        "Python is a popular programming language for web development and data analysis." 
        ]
```

```python
vector_stores = FAISS.from_texts(docs,embedding)
```

```python
multi_retriever = MultiQueryRetriever.from_llm(retriever=vector_stores.as_retriever(),llm=LLM)
```

```python
query = 'Deep Learning'
result = multi_retriever.invoke(query)
```

```python
generated = multi_retriever.llm_chain.invoke(
    query,
    config={}
)
```
