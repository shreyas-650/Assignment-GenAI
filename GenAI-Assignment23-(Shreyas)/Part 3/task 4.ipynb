## Task 4: Maximal Marginal Relevance (MMR) Retriever

```python
from langchain_community.vectorstores import FAISS
from langchain_huggingface import HuggingFaceEmbeddings
```

```python
docs = [ "Solar power converts sunlight into electricity.", 
        "Solar panels are widely used in residential homes.", 
        "Wind turbines harness wind energy for electricity.", 
        "Hydropower generates energy from flowing water.", 
        "Geothermal energy comes from heat inside the Earth."
]
```

```python
embedding = HuggingFaceEmbeddings(model='sentence-transformers/all-MiniLM-L6-v2')
```

```python
vector_store = FAISS.from_texts(docs,embedding)
```

```python
retrieval = vector_store.as_retriever(search_type='mmr',search_kwargs={'k':3,'lambda_mult':0.5})
```

```python
query = 'Renewable Energy'
res = retrieval.invoke(query)
similarity_search_res = vector_store.similarity_search(query,k=3)
```

```python
for i in res:
    print(i.page_content)
```

```python
for i in similarity_search_res:
    print(i.page_content)
```
