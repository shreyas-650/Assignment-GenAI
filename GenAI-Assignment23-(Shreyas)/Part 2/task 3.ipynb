## Task 3: Vector Store Retriever

```python
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_chroma import Chroma
from langchain_huggingface import HuggingFaceEmbeddings
```

```python
docs = TextLoader('../data/obama.md',encoding='utf-8').load()
chunks = RecursiveCharacterTextSplitter(chunk_size=1000).split_documents(docs)
embedding = HuggingFaceEmbeddings(model='sentence-transformers/all-MiniLM-L6-v2')
```

```python
vector_store = Chroma.from_documents(chunks,embedding,collection_name='My_collection')
```

```python
retrieval = vector_store.as_retriever(search_kwargs={"k":3})
```

```python
res = retrieval.invoke('terrorist country')
```
