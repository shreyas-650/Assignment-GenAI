Answer briefly: 
1. Difference between retriever-based RAG and normal prompting?
Retriever based RAG give information of the context we provided whereas in normal prompting it doesn't have our data.
2. Why vector stores are critical?
Vector Stores help to store the chunks data which is critical and it helps in data privacy.
3. When to use MMR vs similarity search?
MMR is used when we want a creative answer. whereas similarity search don't have variety in answer
4. Benefits of multi-query retrieval?
LLM generates multiple questions then it finds the solution for the multiple questions which gives more depth in the answer.
5. Importance of contextual compression?
It removes the noise or unwanted data after the retrieval. Gives only the information we need.
